##Advanced Features of ufuncs

Out Parameter

```python
result = np.empty(3)
np.add(a,b, out = result)
print(result)
```

```python
c = np.array([10,20,30])
np.add(a,b, out = c)
print(c)
```

Where Parameter

```python
a1 = np.array([10,20,30])
b1 = np.array([1,2,3])
d1 = np.add(a1,b1, where = [True, False, False])
print(d1)
```

##np.newaxis

```python
import numpy as np

# Original 1D array
X = np.array([1, 2, 3])       # shape (3,)
print("Original X:", X)
print("Shape of X:", X.shape)
print()

# Add a new axis (convert to column vector)
X2 = X[:, np.newaxis]         # shape (3,1)
print("X after adding newaxis:\n", X2)
print("Shape of X2:", X2.shape)
print()

# Another 1D array (row vector)
Y = np.array([10, 20, 30])    # shape (3,)
print("Y:", Y)
print("Shape of Y:", Y.shape)
print()

# Broadcasting X2 (3,1) + Y (3,) → result shape (3,3)
Z = X2 + Y
print("Result of X2 + Y:\n", Z)
print("Shape of Z:", Z.shape)
```

#4. Custom Functions()

###np.vectorize()

```python
def square(x):
  return x**2

vec_square = np.vectorize(square)

arr = np.array([2,3,5,8])
result = vec_square(arr)
print(result)
```

##Conditional Function (ML style)

```python
def relu(x):
  return x if x > 0 else 0

arr = np.array([-2, -1, 0, 1, 2])
vec_relu = np.vectorize(relu)
result = vec_relu(arr)
print(result)
```

```python
def square(x):
  return x+x

vector = np.vectorize(square)
y = np.array([2,6,8,9])
vectorized = vector(y)
print(vectorized)
```

##np.frompyfunc

```python
def cube(x):
  return x**3
y = np.array([2,3,4])
vector1 = np.frompyfunc(cube,1,1)
vectorized1 = vector1(y)
print(vectorized1)
```

##Broadcasting with Custom ufuncs

```python
def add_broad(x,y):
  return x+y

vector2 = np.vectorize(add_broad)
a = np.array([[2,4,6],[3,5,7]])
b = np.array([10,20,30])
vectorized2 = vector2(a,b)
print(vectorized2)
```

##ML/Feature Engineering Examples

Sigmoid Function

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

vec_sigmoid = np.vectorize(sigmoid)
arr = np.array([0.2, 0.6, 0.4, 0.2])
vectorized_sigmoid = vec_sigmoid(arr)
print(vectorized_sigmoid)
```

Threshold Function

```python
def threshold(x, t=0.5):
  return 1 if x > t else 0
vec_threshold = np.vectorize(threshold)
a = np.array([1, 0.2, 0.4, 0.8, 0.5])
vectorized_threshold = vec_threshold(a)
print(vectorized_threshold)
```
