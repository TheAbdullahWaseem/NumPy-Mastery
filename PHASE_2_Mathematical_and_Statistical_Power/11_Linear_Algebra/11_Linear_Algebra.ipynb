#A. Vector and Matrix Multiplication

Vector and matrix multiplication is one of the most important parts of NumPy linear algebra and is fundamental to Machine Learning.

The main NumPy tools covered in this section are:

- `np.dot()`
- The `@` operator
- `np.matmul()`

Before using them, it is essential to distinguish **element-wise multiplication** from **dot/matrix multiplication**.

## Element-wise Multiplication

The `*` operator performs multiplication element by element.

For two arrays of the same shape:

$
[
C_{ij} = A_{ij}B_{ij}
]
$

For example:

$
[
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix}
*
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}
=
\begin{bmatrix}
1(5) & 2(6)\\
3(7) & 4(8)
\end{bmatrix}
=
\begin{bmatrix}
5 & 12\\
21 & 32
\end{bmatrix}
]
$

## Matrix Multiplication

Matrix multiplication is different. It follows the **row-by-column rule**.

If:

$
[
A \in \mathbb{R}^{m\times n}
]
$

and

$
[
B \in \mathbb{R}^{n\times p}
]
$

then:

$
[
AB \in \mathbb{R}^{m\times p}
]
$

The inner dimensions must match:

$
[
(m\times n)(n\times p) = (m\times p)
]
$

Each element of the resulting matrix is calculated as:

$
[
C_{ij} = \sum_{k=1}^{n} A_{ik}B_{kj}
]
$

This row-by-column operation is the basis of matrix multiplication used throughout Machine Learning.

````python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)
````

### Understanding the Difference

`A * B` multiplies corresponding elements:

$
[
1(5),\; 2(6),\; 3(7),\; 4(8)
]
$

Therefore:

$
[
A*B =
\begin{bmatrix}
5 & 12\\
21 & 32
\end{bmatrix}
]
$

`A @ B` performs matrix multiplication.

The first element is calculated using the first row of `A` and the first column of `B`:

$
[
(1)(5)+(2)(7)=19
]
$

The complete result is:

$
[
AB =
\begin{bmatrix}
19 & 22\\
43 & 50
\end{bmatrix}
]
$

Therefore, always remember:

- `*` → element-wise multiplication
- `@` → matrix multiplication

````python
print("A shape:", A.shape)
print("B shape:", B.shape)
print("A @ B shape:", (A @ B).shape)
````

### Understanding the Shapes

Here:

$
[
A = (2,2)
]
$

and:

$
[
B = (2,2)
]
$

Therefore:

$
[
(2\times2)(2\times2)=(2\times2)
]
$

The two inner dimensions are both `2`, so matrix multiplication is valid.

A useful rule is:

$
[
(m\times n)(n\times p)=(m\times p)
]
$

The `n` values must be equal.

The resulting matrix gets the **outer dimensions**: `m × p`.

When working with NumPy, a shape mismatch produces an error rather than automatically performing the multiplication.

````python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [1, 2],
    [3, 4]
])

#A @ B
````

A useful rule to check before matrix multiplication is:

$
[

#B. Matrix Inversion with `np.linalg.inv()`

Matrix inversion is an important linear algebra operation used in Machine Learning, particularly when working with systems of equations and the mathematical formulation of Linear Regression.

For a square matrix $A$, its inverse is written as $A^{-1}$.

The defining property of an inverse is:

$
AA^{-1}=A^{-1}A=I
$

where $I$ is the **identity matrix**.

The identity matrix is the matrix equivalent of the number `1`. For example:

$
I=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
$

Therefore, multiplying a matrix by its inverse gives the identity matrix:

$
AA^{-1}=I
$

An inverse exists only for a **square, non-singular matrix**.

In NumPy, the inverse of a matrix is calculated using:

`np.linalg.inv()`

````python
A = np.array([
    [4, 7],
    [2, 6]
])

A_inv = np.linalg.inv(A)

print(A_inv)
````

The result is the inverse matrix $A^{-1}$.

For this matrix:

$
A=
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
$

NumPy calculates:

$
A^{-1}=
\begin{bmatrix}
0.6 & -0.7\\
-0.2 & 0.4
\end{bmatrix}
$

You normally do not calculate the inverse manually when working with NumPy. The purpose here is to understand what `np.linalg.inv()` is producing.

````python
identity = A @ A_inv
print(identity)
````

The result is approximately the identity matrix:

$
AA^{-1}=I
$

Due to floating-point arithmetic, NumPy may display very small values such as `1.11022302e-16` instead of exactly `0`. These values are numerical precision errors and are effectively zero for this calculation.

A better way to verify the result is to use `np.allclose()`.

````python
print(np.allclose(A @ A_inv, np.eye(2)))
````

`np.allclose()` checks whether two arrays are numerically close to each other.

`np.eye(2)` creates a $2\times2$ identity matrix:

$
I=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
$

Therefore, the expression checks whether:

$
AA^{-1}\approx I
$

and returns `True` if the matrices are sufficiently close.

# When Does a Matrix Have an Inverse?

A matrix must satisfy two important conditions:

1. It must be **square**.

$
(n\times n)
$

2. It must be **non-singular**, meaning its determinant is not zero.

$
\det(A)\neq0
$

If:

$
\det(A)=0
$

the matrix is called **singular** and does not have an ordinary inverse.

For example, the following matrix is singular because its rows are linearly dependent.

````python
A = np.array([
    [1, 2],
    [2, 4]
])

print(np.linalg.det(A))
````

The determinant is zero:

$
\det(A)=0
$

Therefore, this matrix has no inverse.

Attempting to use `np.linalg.inv()` on it will produce a `LinAlgError`.

````python
# This will raise an error because A is singular.

# A_inv = np.linalg.inv(A)
````

# Why Matrix Inversion Matters in Machine Learning

Matrix inversion appears in the mathematical solution of Ordinary Least Squares Linear Regression.

One form of the normal equation is:

$
\hat{\beta}=(X^TX)^{-1}X^Ty
$

Here:

- $X$ is the feature matrix.
- $y$ is the target vector.
- $\beta$ represents the model coefficients.
- $X^T$ is the transpose of $X$.
- $(X^TX)^{-1}$ is the inverse of the resulting matrix.

This is an important connection between NumPy linear algebra and Machine Learning.

However, in practical numerical ML code, we generally avoid explicitly calculating an inverse when we only need to solve a system of equations. `np.linalg.solve()` is usually preferred for that purpose because it is more numerically appropriate and efficient.

Therefore, `np.linalg.inv()` is important to understand, but you should not automatically use it whenever an equation contains an inverse.

The **determinant** is a single number calculated from a square matrix.

In NumPy, the determinant is calculated using:

$

$

Therefore:

$
\det(A)=(4)(6)-(7)(2)
$

$
=24-14
$

$
=10
$

Since the determinant is non-zero, the matrix is non-singular and has an inverse.

````python
A_inv = np.linalg.inv(A)

print(A_inv)
````

The inverse exists because:

$
\det(A)=10\neq0

This demonstrates the relationship:

$

#C. Eigenvalues and Eigenvectors with `np.linalg.eig()`

Eigenvalues and eigenvectors describe important properties of a matrix and are especially important in **Machine Learning, PCA, dimensionality reduction, covariance analysis, and linear transformations**.

For a square matrix $A$, an eigenvector $v$ and its corresponding eigenvalue $\lambda$ satisfy:

$
Av=\lambda v
$

This means that when matrix $A$ transforms the vector $v$, the **direction of $v$ does not change**. Only its magnitude (and possibly its sign) changes.

- $A$ → square matrix
- $v$ → eigenvector
- $\lambda$ → eigenvalue

The NumPy function used to calculate them is:

```python
np.linalg.eig(A)

````python

import numpy as np

A = np.array([
    [4, 1],
    [2, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
````

## Understanding the Output

`np.linalg.eig()` returns two arrays:

1. **Eigenvalues** — one value for each eigenvector.
2. **Eigenvectors** — stored as columns of the returned matrix.

For example, if:

$
\lambda_1,\lambda_2
$

are the eigenvalues, the corresponding eigenvectors are the columns:

$
V=[v_1\quad v_2]
$

So:

- `eigenvalues[0]` corresponds to `eigenvectors[:, 0]`
- `eigenvalues[1]` corresponds to `eigenvectors[:, 1]`

The ordering of eigenvalues/eigenvectors should not be assumed to have a particular meaning; what matters is that each eigenvalue matches its corresponding eigenvector.

## Verify an Eigenvalue-Eigenvector Pair

Suppose `eigenvalues[0]` is $\lambda$ and `eigenvectors[:, 0]` is $v$.

The defining relationship is:

$
Av=\lambda v
$

We can verify this directly with NumPy.

````python
A = np.array([
    [4, 1],
    [2, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(A)


v = eigenvectors[:, 0]
lam = eigenvalues[0]

print("A @ v:")
print(A @ v)

print("\nλ * v:")
print(lam * v)

print("\nVerified:")
print(np.allclose(A @ v, lam * v))
````

The final result should be `True`.

This is the most important idea to remember:

$
Av=\lambda v
$

The matrix multiplication changes the vector's magnitude according to $\lambda$, while the vector remains in the same direction.

## Eigenvalues and Eigenvectors Are Paired

If:

```python
eigenvalues[0]

is an eigenvalue, its corresponding eigenvector is:

\begin{bmatrix}

Similarly:

This makes the meaning of an eigenvalue very intuitive: it tells us the scaling factor applied to its corresponding eigenvector.

````python
A = np.array([
    [2, 0],
    [0, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
````

#D. Solving Systems of Linear Equations with `np.linalg.solve()`

A **system of linear equations** contains multiple equations whose unknown variables must satisfy all equations simultaneously.

For example:

$
2x+y=8
$

$
x+3y=13
$

We can write the system in matrix form as:

$
Ax=b
$

where:

- $A$ → coefficient matrix
- $x$ → vector of unknown variables
- $b$ → vector of known results

For the example:

$
A=
\begin{bmatrix}
2&1\\
1&3
\end{bmatrix},
\quad
x=
\begin{bmatrix}
x\\
y
\end{bmatrix},
\quad
b=
\begin{bmatrix}
8\\
13
\end{bmatrix}
$

NumPy provides:

```python
np.linalg.solve(A, b)

````python
import numpy as np

A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([8, 13])

x = np.linalg.solve(A, b)

print(x)
````

The result is:

$
x=
\begin{bmatrix}
2.2\\
3.6
\end{bmatrix}
$

So:

$
x=2.2,\qquad y=3.6
$

We can verify the solution by substituting it back into the original system:

$
Ax=b
$

If the solution is correct, multiplying `A @ x` should give `b`.

````python
print("A @ x:")
print(A @ x)

print("\nb:")
print(b)

print("\nVerified:")
print(np.allclose(A @ x, b))
````

## Why `np.linalg.solve()` Instead of `np.linalg.inv()`?

You may remember the relationship:

$
Ax=b
$

Mathematically, if $A$ is invertible, we could write:

$
x=A^{-1}b
$

However, in NumPy you generally should **not explicitly calculate the inverse** just to solve a system.

Instead of:

```python
x = np.linalg.inv(A) @ b

prefer:

``
x = np.linalg.solve(A, b)
``

`np.linalg.solve()` is designed specifically for solving linear systems and is generally more numerically appropriate and efficient than explicitly computing the inverse.

## A Larger System

The same idea works with three or more unknowns.

Consider:

$
2x+y+z=9
$

$
x+3y+2z=13
$

$
3x+2y+4z=24
$

Write it as:

$
Ax=b
$

where:

$
A=
\begin{bmatrix}

1&3&2\\
3&2&4
\end{bmatrix}
$

and:

$
b=
\begin{bmatrix}
9\\
13\\
24
\end{bmatrix}
$

````python
A = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [3, 2, 4]
])

b = np.array([9, 13, 24])

x = np.linalg.solve(A, b)

print(x)

print("\nVerification:")
print(np.allclose(A @ x, b))
````

## Shape Requirements

For:

$
Ax=b
$

if $A$ has shape $(n,n)$, then $b$ normally has shape $(n,)$ or $(n,k)$.

For example:

```python

# (3, 3)

b.shape
# (3,)

The number of rows in A must match the number of elements/rows in b.

A must be a square, non-singular matrix for np.linalg.solve() to produce a unique solution.

If the coefficient matrix is singular, NumPy raises a LinAlgError.

````python
A = np.array([
    [1, 2],
    [2, 4]
])

b = np.array([5, 10])

try:
    x = np.linalg.solve(A, b)
    print(x)
except np.linalg.LinAlgError as e:
    print("Error:", e)
````

````python
print(B.shape)
````

Each column of B represents a different right-hand side, and each corresponding column of X contains its solution.

This is useful when the same coefficient matrix is used repeatedly with different target values.

we can conceptually solve the system:

using a linear-system solver.

In practical ML libraries, specialized algorithms are often preferred, but understanding np.linalg.solve() gives you the underlying mathematical foundation for many optimization and regression methods.

#E. Matrix Rank and Its Importance for Independent Information and Model Stability

The **rank of a matrix** tells us how much **independent information** the matrix contains.

More precisely, the rank is the number of **linearly independent rows or columns**.

In NumPy:

```python
np.linalg.matrix_rank(A)

````python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

rank = np.linalg.matrix_rank(A)

print(rank)
````

## Understanding Rank

Consider:

$
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
$

The two rows are independent, and the two columns are also independent.

Therefore:

$
\operatorname{rank}(A)=2
$

A $2\times2$ matrix can have a maximum rank of $2$.

In general:

$
\operatorname{rank}(A)\leq\min(m,n)
$

for a matrix with shape $m\times n$.

## A Matrix with Redundant Information

Consider:

$
B=
\begin{bmatrix}
1&2\\
2&4
\end{bmatrix}
$

The second row is simply:

$
2\times[1,2]=[2,4]
$

Therefore, the second row does not provide new independent information.

The matrix has only **one independent row**:

$
\operatorname{rank}(B)=1
$

````python
B = np.array([
    [1, 2],
    [2, 4]
])

print(np.linalg.matrix_rank(B))
````

## Full Rank vs Rank Deficient

For a square $n\times n$ matrix:

**Full rank**

$
\operatorname{rank}(A)=n
$

means all rows/columns contain independent information.

**Rank deficient**

$
\operatorname{rank}(A)<n
$

means some rows or columns are linearly dependent.

For example:

$
A=
\begin{bmatrix}
1&2\\
2&4
\end{bmatrix}
$

has rank $1$ instead of $2$, so it is rank deficient.

This is closely related to the determinant and inverse:

$
\operatorname{rank}(A)=n
\Longleftrightarrow
\det(A)\neq0
$

for a square matrix.

Therefore, a full-rank square matrix is invertible, while a rank-deficient square matrix is singular.

## Rank and Features in Machine Learning

Suppose a dataset contains three features:

```text
Feature 1
Feature 2
Feature 3

If:

$
Feature_3=Feature_1+Feature_2
$

then Feature 3 does not provide completely new information.

For example:

$
X=
\begin{bmatrix}
1&2&3\\
2&4&6\\
3&1&4
\end{bmatrix}
$

The third column is the sum of the first two columns:

$
3=1+2
$

$
6=2+4
$

$
4=3+1
$

Therefore, the columns are linearly dependent and the matrix cannot have full column rank.

````python
X = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 1, 4]
])

print("Shape:", X.shape)
print("Rank:", np.linalg.matrix_rank(X))
````

## Why This Matters for Machine Learning

Rank tells us whether our features contain enough **independent information**.

Suppose a feature matrix $X$ has $p$ features.

If:

$
\operatorname{rank}(X)<p
$

then at least one feature can be represented as a linear combination of other features.

This is a form of **perfect multicollinearity**.

For example:

$
X_3=X_1+X_2
$

means that knowing $X_1$ and $X_2$ already determines $X_3$.

The third feature therefore does not add independent information.

This can create problems when algorithms require operations involving:

$
X^TX
$

because a rank-deficient feature matrix can make $X^TX$ singular and therefore non-invertible.

## Rank and Model Stability

Rank is also connected to **model stability**.

Consider the linear regression expression:

$
\beta=(X^TX)^{-1}X^Ty
$

For this ordinary inverse to exist, $X^TX$ must be full rank.

If features are perfectly dependent, $X^TX$ becomes singular and the ordinary inverse does not exist.

Even when features are not perfectly dependent but are **very strongly correlated**, the matrix can become **nearly singular**.

This can make coefficient estimates sensitive to small changes in the data.

Therefore:

- Full rank → independent information is available.
- Rank deficient → redundant information exists.
- Nearly rank deficient → potential numerical instability.

````python
# A simple feature matrix

X = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9],
    [4, 8, 12]
])

rank = np.linalg.matrix_rank(X)

print("Shape:", X.shape)
print("Rank:", rank)
````

The matrix has three columns but its rank is only one.

Why?

Because:

$
X_2=2X_1
$

and:

$
X_3=3X_1
$

So although there are **3 columns**, there is only **1 independent direction of information**.

This distinction is important:

**Number of features ≠ amount of independent information.**

## Rank Is Not the Same as Number of Rows or Columns

Consider a matrix with shape `(1000, 20)`.

It contains:

- 1000 rows
- 20 columns

But its rank can be anywhere from $0$ to $20$.

For a feature matrix:

$
X\in\mathbb{R}^{1000\times20}
$

the maximum possible rank is:

$
\operatorname{rank}(X)\leq20
$

If the rank is `20`, all 20 features contribute independent linear directions.

If the rank is `15`, only 15 independent directions exist among those 20 features.

## Practical NumPy Check

When working with a dataset, you can quickly inspect the relationship between dimensions and rank:

```python
X.shape
np.linalg.matrix_rank(X)

````python
X = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
])

print("Shape:", X.shape)
print("Rank:", np.linalg.matrix_rank(X))
````

- A useful mental model is:

- Shape tells you how much data is present.

- Rank tells you how much of that data is linearly independent.

## Key Takeaways

- `np.linalg.matrix_rank(A)` calculates the rank of a matrix.
- Rank = number of linearly independent rows or columns.
- Rank cannot exceed the smaller dimension of the matrix:

$
\operatorname{rank}(A)\leq\min(m,n)
$

- **Full rank** means the maximum possible independent information is present.
- **Rank deficient** means some information is linearly redundant.
- In ML, rank is important for detecting **feature redundancy and multicollinearity**.
- Rank deficiency can make $X^TX$ singular and cause problems with ordinary matrix inversion.
- Nearly dependent features can contribute to **numerical/model instability**, even when the matrix technically has full rank.
- Therefore, matrix rank connects linear algebra directly to **feature engineering, regression, dimensionality reduction, and numerical stability**.

#F. Vector and Matrix Norms — Measuring Magnitude and Distance

A **norm** is a function that measures the magnitude or size of a vector or matrix.

In Data Science and Machine Learning, norms are commonly used for:

- measuring vector magnitude
- measuring distance between points
- measuring prediction error
- regularization
- optimization
- measuring the size of model parameters

NumPy provides:

```python
np.linalg.norm()

````python
import numpy as np

v = np.array([3, 4])

print(np.linalg.norm(v))
````

## L2 Norm — Euclidean Magnitude

The L2 norm is the ordinary geometric length of a vector.

For:

$
v=[v_1,v_2,\ldots,v_n]
$

the L2 norm is:

$
\|v\|_2=\sqrt{v_1^2+v_2^2+\cdots+v_n^2}
$

For $v=[3,4]$:

$
\|v\|_2=\sqrt{3^2+4^2}=5
$

In NumPy, L2 is the default vector norm:

```python
np.linalg.norm(v)

or explicitly

````python
print(np.linalg.norm(v, ord = 2))
````

````python
v = np.array([3, 4])

print("L2 norm:", np.linalg.norm(v))
print("Explicit L2:", np.linalg.norm(v, ord=2))
````

## L1 Norm — Sum of Absolute Values

The L1 norm adds the absolute values of all elements:

$
\|v\|_1=\sum_{i=1}^{n}|v_i|
$

For:

$
v=[-3,4,-2]
$

we get:

$
\|v\|_1=|-3|+|4|+|-2|=9
$

In NumPy:

```python
np.linalg.norm(v, ord=1)

L1 norms are particularly important in ML because L1 regularization can encourage model coefficients to become exactly zero, which can help with feature selection.

````python
v = np.array([-3, 4, -2])

print("L1 norm:", np.linalg.norm(v, ord=1))
print("L2 norm:", np.linalg.norm(v, ord=2))
````

## L1 vs L2 in Machine Learning

Suppose a model has parameter vector:

$
w=[2,-1,0.5,-3]
$

The L1 penalty is:

$
\|w\|_1
$

while the L2 penalty is:

$
\|w\|_2^2
$

These appear in regularization:

**L1 regularization — Lasso**

$
Loss+\lambda\|w\|_1
$

**L2 regularization — Ridge**

$
Loss+\lambda\|w\|_2^2
$

The purpose is to penalize excessively large model parameters and help control model complexity.

The important distinction is that L1 regularization tends to produce sparse coefficients, while L2 generally shrinks coefficients toward zero without typically making them exactly zero.

````python
w = np.array([2, -1, 0.5, -3])

l1 = np.linalg.norm(w, ord=1)
l2 = np.linalg.norm(w, ord=2)

print("L1 norm:", l1)
print("L2 norm:", l2)
print("Squared L2 norm:", l2 ** 2)
````

## Norm as Distance Between Data Points

The L2 norm can also measure the distance between two vectors.

Given two points $x$ and $y$:

$
d(x,y)=\|x-y\|_2
$

For example:

$
x=[1,2,3]
$

$
y=[4,6,3]
$

First calculate the difference:

$
x-y=[-3,-4,0]
$

Then:

$
\|x-y\|_2=5
$

This is the standard **Euclidean distance** used in many ML algorithms, including K-Nearest Neighbors and clustering.

````python
x = np.array([1, 2, 3])
y = np.array([4, 6, 3])

distance = np.linalg.norm(x - y)

print("Euclidean distance:", distance)
````

## Matrix Norms

`np.linalg.norm()` can also measure the magnitude of a matrix.

For a matrix, the default norm is the **Frobenius norm**:

$
\|A\|_F=
\sqrt{\sum_{i,j}|A_{ij}|^2}
$

It is essentially the L2 magnitude of all matrix elements treated collectively.

For:

$
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
$

the Frobenius norm is:

$
\|A\|_F=\sqrt{1^2+2^2+3^2+4^2}
$


````python
A = np.array([
    [1, 2],
    [3, 4]
])

print("Frobenius norm:", np.linalg.norm(A))
````

## Other Matrix Norms

You can specify different norms using `ord`.

### Matrix L1 norm

For matrices, `ord=1` gives the **maximum absolute column sum**.

### Matrix L∞ norm

`ord=np.inf` gives the **maximum absolute row sum**.

These are useful when analyzing numerical behavior and the magnitude of transformations.

````python
A = np.array([
    [1, -2],
    [3, 4]
])

print("Frobenius norm:", np.linalg.norm(A))
print("Matrix L1 norm:", np.linalg.norm(A, ord=1))
print("Matrix L-infinity norm:", np.linalg.norm(A, ord=np.inf))
````

## Norms in Optimization

Norms appear constantly in optimization because they provide a mathematical way to measure:

- parameter magnitude
- distance from a target
- error size
- regularization strength

For example, if a prediction vector is $\hat{y}$ and the actual values are $y$, the error vector is:

$
e=y-\hat{y}
$

Its L2 norm:

$
\|e\|_2
$

measures the overall magnitude of the prediction error.

This is one reason norms are fundamental to the mathematics behind ML loss functions and optimization algorithms.

## Key Takeaways

- `np.linalg.norm()` measures the magnitude of vectors or matrices.
- **L1 norm** → sum of absolute values.
- **L2 norm** → Euclidean magnitude.
- **Frobenius norm** → common matrix magnitude measure.
- $\|x-y\|_2$ gives Euclidean distance between two vectors.
- L1 and L2 norms are fundamental to **regularization and optimization**.
- Norms let ML algorithms quantify **size, distance, error, and model complexity**.

#G. Shape, Dimensionality, and Compatibility in Linear Algebra

In NumPy linear algebra, many errors come from one thing: **incompatible shapes**.

Before performing an operation, always ask:

1. What are the shapes of the arrays?
2. Is the operation compatible with those shapes?
3. What will the resulting shape be?

This is especially important for Machine Learning because datasets, feature matrices, parameter vectors, and target vectors constantly interact through matrix operations.

````python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A.shape)
print(A.ndim)
````

`A.shape` is `(2, 3)`:

- 2 rows
- 3 columns

`A.ndim` is `2` because `A` is a 2-dimensional array.

Remember:

```python
A.shape


## 1. Vector Shapes

NumPy commonly represents a vector as a 1D array:

```python
v = np.array([1, 2, 3])

`(3,)`

This is not the same as a column matrix with shape (3, 1).

A 1D vector:

$
v\in\mathbb{R}^{3}
$

A column vector:

$
v\in\mathbb{R}^{3\times1}
$

A row vector:

$
v^T\in\mathbb{R}^{1\times3}
$

This distinction becomes important when performing matrix multiplication.

````python

v = np.array([1, 2, 3])

column_v = v.reshape(3, 1)
row_v = v.reshape(1, 3)

print("v:", v.shape)
print("column_v:", column_v.shape)
print("row_v:", row_v.shape)
````

## 2. Matrix Multiplication Compatibility

For:

$
A\in\mathbb{R}^{m\times n}
$

and:

$
B\in\mathbb{R}^{n\times p}
$

matrix multiplication is valid because the **inner dimensions match**:

$
(m\times n)(n\times p)
$

The result has shape:

$
m\times p
$

The rule is therefore:

**columns of the first matrix = rows of the second matrix.**

````python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])                    # (2, 3)

B = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])                    # (3, 2)

C = A @ B

print("A:", A.shape)
print("B:", B.shape)
print("C:", C.shape)
````

Here:

$
(2\times3)(3\times2)=(2\times2)
$

So the multiplication is valid and the result has shape `(2, 2)`.

If the shapes were `(2, 3)` and `(2, 2)`, multiplication would be invalid because:

$
(2\times3)(2\times2)
$

has mismatched inner dimensions `3` and `2`.

## 3. Matrix-Vector Multiplication

Suppose:

$
X\in\mathbb{R}^{m\times n}
$

and:

$
w\in\mathbb{R}^{n}
$

Then:

$
Xw\in\mathbb{R}^{m}
$

For example:

$
(3\times2)(2)=(3)
$

This is one of the most important shape relationships in Machine Learning because a feature matrix is commonly multiplied by a parameter vector.

````python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

w = np.array([10, 20])

result = X @ w

print("X shape:", X.shape)
print("w shape:", w.shape)
print("Result shape:", result.shape)
print(result)
````

## 4. Why `(n,)` and `(n,1)` Are Different

Consider:

w = np.array([10, 20])

Its shape is `(2,)`.

But:

`w = np.array([[10], [20]])`

has shape `(2, 1)`.

Both contain two values, but NumPy treats them differently during linear algebra operations.

For example:

$
(3\times2)(2)=(3)
$

while:

$
(3\times2)(2\times1)=(3\times1)
$

So the resulting shapes are different.

````python
w = np.array([[10], [20]])
print(w)
print(w.shape)
````

````python

X = np.ones((3, 2))

w_1d = np.array([10, 20])
w_column = np.array([
    [10],
    [20]
])

print("X @ w_1d:", (X @ w_1d).shape)
print("X @ w_column:", (X @ w_column).shape)
````

## 5. Transpose Changes Shape

For a 2D matrix:

$
A\in\mathbb{R}^{m\times n}
$

its transpose has shape:

$
A^T\in\mathbb{R}^{n\times m}
$

This is often used to make matrix multiplication compatible.

For example:

$
X\in\mathbb{R}^{100\times5}
$

Then:

$
X^T\in\mathbb{R}^{5\times100}
$

Therefore:

$
X^TX
$

has shape:

$
(5\times100)(100\times5)=(5\times5)
$

This is the shape of the feature-feature matrix that appears in linear regression mathematics.

````python
X = np.ones((100, 5))

print("X:", X.shape)
print("X.T:", X.T.shape)
print("X.T @ X:", (X.T @ X).shape)
````

## 6. `np.dot()` and `@` Need Careful Shape Reasoning

For the 2D matrices normally used in ML:

```python

`np.matmul(A, B)`

For 1D arrays, `np.dot()` behaves differently from ordinary matrix notation because a `1D × 1D` operation produces a scalar dot product.

For example:

$
(3)\cdot(3)\rightarrow\text{scalar}
$

Therefore, always check whether your arrays are 1D or 2D before reasoning about the result.

````python

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a shape:", a.shape)
print("b shape:", b.shape)
print("dot result:", np.dot(a, b))
print("result shape:", np.dot(a, b).shape)
````

## 7. Requirements for `np.linalg.solve()`

For:

$
Ax=b
$

`np.linalg.solve(A, b)` expects a square coefficient matrix.

If:

A\in\mathbb{R}^{n\times n}

$
b\in\mathbb{R}^{n}

For example:

(3\times3)(3)=(3)

````python
A = np.eye(3)

b = np.array([1, 2, 3])

B = np.array([
    [1, 10],
    [2, 20],
    [3, 30]
])

print("solve(A, b):", np.linalg.solve(A, b).shape)
print("solve(A, B):", np.linalg.solve(A, B).shape)
````

## 9. Requirements for `np.linalg.eig()`

`np.linalg.eig(A)` operates on a square matrix:

$
A\in\mathbb{R}^{n\times n}
$

It returns:

- `eigenvalues` → shape `(n,)`
- `eigenvectors` → shape `(n, n)`

Each eigenvector is stored as a column.

So:

eigenvectors[:, i]

## 10. Rank and Shape

For:

$
A\in\mathbb{R}^{m\times n}
$

the rank can never exceed either dimension:

$
\operatorname{rank}(A)\leq\min(m,n)
$

For example, a matrix with shape `(100, 5)` can have a maximum rank of `5`.

If its rank is `3`, then only three independent linear directions exist among its five columns.

This is particularly important when interpreting a feature matrix: **five features do not necessarily mean five independent pieces of information.**

## 11. A Practical ML Shape Pattern

A very common ML setup is:

$
X\in\mathbb{R}^{m\times n}
$

where:

- $m$ = number of observations
- $n$ = number of features

and:

$
w\in\mathbb{R}^{n}
$

Then:

$
Xw\in\mathbb{R}^{m}
$

If we add an intercept:

$
\hat{y}=Xw+b
$

the prediction vector still has shape:

$
\hat{y}\in\mathbb{R}^{m}
$

This shape reasoning is essential when implementing models manually with NumPy.

````python
# Example: 4 observations and 3 features

X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6]
])

w = np.array([0.5, 1.0, 2.0])
b = 3

y_hat = X @ w + b

print("X:", X.shape)
print("w:", w.shape)
print("y_hat:", y_hat.shape)
print(y_hat)
````

## Shape Compatibility Cheat Sheet

| Operation | Typical requirement | Result |
|---|---|---|
| `A @ B` | `(m, n) @ (n, p)` | `(m, p)` |
| `A @ v` | `(m, n) @ (n,)` | `(m,)` |
| `A @ v_column` | `(m, n) @ (n, 1)` | `(m, 1)` |
| `A.T` | `(m, n)` | `(n, m)` |
| `np.linalg.inv(A)` | square `(n, n)` | `(n, n)` |
| `np.linalg.det(A)` | square `(n, n)` | scalar |
| `np.linalg.solve(A, b)` | `A: (n,n)`, `b: (n,)` | `(n,)` |
| `np.linalg.eig(A)` | square `(n,n)` | values `(n,)`, vectors `(n,n)` |
| `np.linalg.matrix_rank(A)` | `(m,n)` | scalar |

The most important rule to remember for matrix multiplication is:

$
(m\times n)(n\times p)\rightarrow(m\times p)
$

**Inner dimensions must match; outer dimensions determine the result.**

## Key Takeaways

- `shape` tells you the size of each dimension.
- `ndim` tells you the number of dimensions.
- A 1D vector `(n,)` is different from a column vector `(n,1)`.
- For matrix multiplication, the inner dimensions must match:

$
(m\times n)(n\times p)\rightarrow(m\times p)
$

- `inv`, `det`, `solve`, and `eig` have specific shape requirements.
- `X @ w` is one of the most important shape patterns in ML:

$
(m\times n)(n)\rightarrow(m)
$

- Always check `.shape` before debugging a linear algebra operation.
- Shape reasoning is not just NumPy syntax; it is how you understand whether the underlying linear algebra is mathematically valid.
