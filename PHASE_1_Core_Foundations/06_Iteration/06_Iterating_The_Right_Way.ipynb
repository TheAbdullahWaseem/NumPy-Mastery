##np.nditer()

```python
a1 = np.array([[1,2],[3,4]])
print(a1)
```

```python
for x in a1:
  print(x)
```

```python
for x in np.nditer(a1):
  print(x)
```

```python
for x in np.nditer(a1, op_flags= ['readwrite']):
  x[...] = x * 2
print(a1)
```

```python
for x, y in np.nditer([h,i]):
  print(x,y)
```

##Practice Problems
Replace all negative numbers with 0.

```python
a5 = np.arange(24).reshape(2,3,4)
a5[0,1,1:3] = -4
a5[1,0:2,1] = -67
print(a5)
```

```python
a5[a5 < 0] = 0
print(a5)
```

Replace even numbers with square



```python
mask = a5 %2 == 0
a5[mask] = a5[mask] ** 2
print(a5)
```

Multiply each row of 2D array by [1,2,3].

```python
t = np.arange(6).reshape(2,3)
print(t)
```

```python
t = t*np.array([1,2,3])
print(t)
```

```python
for x in np.nditer(t, op_flags=['readwrite']):
  x[...] = x * [1,2,3]
print(t)
```
