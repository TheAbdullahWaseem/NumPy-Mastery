```python
import numpy as np
```

```python
arr = np.arange(1,10)
for i in range(len(arr)):
    arr[i] = arr[i] * 2
print(arr)
```

np.nditer()

```python
arr = np.array([[1,2,3],
                [4,5,6]])
for x in np.nditer(arr):
  print(x)
```

```python
c = np.array([5,2,7,2])
for x in np.nditer(c, op_flags=['readwrite']):
  x = x*2
  print(x)
print(c)
```

```python
x = np.array([1,2,3,4,5])
y = np.array([6,7,8,9,10])
for a,b in np.nditer([x,y]):
  print(a,b)
```

```python
arr = np.array([[1,2,3],
                [4,5,6]])

for x in np.nditer(arr):
    print(x, end=" ")
```

```python
arr = np.array([1,2,3,4])

for x in np.nditer(arr, op_flags=['readwrite']):
    x = x * 3

print(arr)
```

```python
arr = np.arange(1,7).reshape(2,3).T

for x in np.nditer(arr):
    print(x, end=" ")
```

```python
a = np.array([1,2,3])
b = np.array([[10],
              [20],
              [30]])

for x, y in np.nditer([a, b]):
    print(x, y)
```

```python
a = np.array([1,2,3])
b = np.array([10,20,30])

for x, y in np.nditer([a,b], op_flags=[['readwrite'], ['readonly']]):
    x[...] = x + y

print(a)
print(b)
```

```python
arr = np.array([1,2,3,4])

it = np.nditer(arr, op_flags=['readwrite'])

for x in it:
    arr[0] = 999
    x[...] = x * 2

print(arr)
```
