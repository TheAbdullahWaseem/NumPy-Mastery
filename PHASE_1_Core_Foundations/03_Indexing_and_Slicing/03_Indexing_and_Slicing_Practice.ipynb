Q4 — Boolean Indexing

arr = np.array([[5,10,15],
                [20,25,30]])

Tasks:

Select all elements greater than 15.

Replace all elements less than 20 with 0.

```python
arr2 = np.array([[5,10,15], [20,25,30]])
print(arr2)
```

```python
print(arr2[arr2 > 15])
```

```python
elements = arr2 < 20
arr2[elements] = 0
print(arr2)
```

Q6 — Boolean + Fancy Combined

arr = np.array([5,10,15,20,25,30])

Task:

Select elements greater than 10 and at odd indices.

```python
arr4 = np.array([5,10,15,20,25,30])
print(arr4)
```

```python
indices = arr4[(arr4 > 10) & (arr4 % 2 == 1)]
print(indices)
```

Q7 — Multidimensional Selection & Broadcasting

arr = np.arange(1,17).reshape(4,4)

Tasks:

Select rows 0 and 2 and columns 1 and 3 using np.ix_().

Select rows 1 and 3 and columns 0 and 2 without np.ix_().

Use ellipsis ... to select first column of all rows.

```python
arr5 = np.arange(1,17).reshape(4,4)
print(arr)
```

Multidimensional Selection (Grid)

Select a 2×2 submatrix across both matrices using np.ix_():

Rows: 0 and 2

Columns: 1 and 3

```python
b2 = np.arange(1, 25).reshape(2,3,4)
print(b2)
```

```python
m1 = b2[0][np.ix_([0,2],[1,3])]
m2 = b2[1][np.ix_([0,2],[1,3])]
print("first_matrix: ",m1)
print("second_matrix: ",m2)
```
