#PART 1 — np.where()
1. Use np.where() to find indices where values are greater than 10.

```python
A = np.array([5, 12, 7, 18, 3, 21])
print(np.where(A>8))
```

2. Replace all values less than 10 with 0 using np.where().

```python
replaced = np.where(A<10,0,A)
print(replaced)
```

3. Without using loops, create a new array where:

even numbers → multiply by 2

odd numbers → keep same

```python
even_odd = np.where(A%2==0,A*2,A)
print(even_odd)
```

##Question 2
1. Find row and column indices where values are divisible by 5 AND greater than 40.

```python
A = np.array([[10, 25, 30],
              [45, 50, 65],
              [70, 85, 90]])
print(A)
print(np.where((A%5==0) & (A>40)))
```

2. Replace elements that are:

  less than row mean

  with -1

```python
row_mean = np.mean(A,axis=1).reshape(-1,1)
print(row_mean)
print(np.where(A<row_mean,-1,A))
```

#PART 6 — Boolean Filtering Mastery
##Question 10
1. Extract elements:

greater than 30

divisible by 3

NOT greater than 80


```python
A = np.array([
    [12, 45, 78],
    [23, 56, 89],
    [34, 67, 90]
])
masked = A[(A>30)&(A%3==0)&(A<80)]
print(masked)
```


2. Replace elements that are:

between 40 and 80

with their square

```python
elements = (A>40) & (A<80)
print(elements,"\n")
A[elements] = A[elements]*2
print(A)
```


3. Extract rows where:

row sum > 150

```python
A = np.array([
    [12, 45, 78],
    [23, 56, 89],
    [34, 67, 90]
])
masking = np.sum(A,axis=1)
masking = masking > 150
print(masking)
A = A[masking]
print(A)
```
