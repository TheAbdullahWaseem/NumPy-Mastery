```python
import numpy as np
```

```python
a = np.arange(1,10)
print(a)
a1 = np.array([[1,2,15,7,8],[10,3,2,12,2]])
print(a1,'\n')
a2 = np.array([[[1,2,3,12,14],[12,4,11,56,1]],[[3,22,5,32,1],[32,2,54,33,2]]])
print(a2)
```

##np.where()

```python
print(np.where(a > 5))
print(np.where(a1>9))
print(np.where(a2>10))
```

##np.where() with condition

```python
print(np.where(a%2==0, a, -1))
print(np.where(a1%2==0, a1, -1))
```

##np.searchsorted()

```python
a3 = np.arange(1,7).reshape(2,3)
print(a3)
print(np.searchsorted(a, 3))
print(np.searchsorted(a, 3, side = 'right'))

#print(np.searchsorted(a3,3)) not work because work only for 1d array
print(np.searchsorted(a3.ravel(),3)) #here we flatten it then it work
#print(np.searchsorted(a1,8)) not work because a1 is not sorted array
```

##np.unique()

```python
a4 = np.array([1,3,3,4,4,4,5,8,19,20,24,25,33,35,55,56])
a5 = np.array([3,4,5,6,2,7,8,9])
print(a4)
print(a5)
```

```python
print(np.unique(a4))
print(np.unique(a5))
print(np.unique(a4, return_counts=True))
print(np.unique(a4, return_index=True))
print(np.unique(a4, return_inverse=True))
```

##np.sort()

```python
a6 = np.array([[1,12,34,4,10,7],[33,5,67,2,21,40]])
print(a6)
```

```python
print(np.sort(a4))
print(np.sort(a6),'\n')
print(np.sort(a6,axis=0),'\n') #column wise
print(np.sort(a6,axis=1)) #row wise
print(np.sort(a6,axis=-1)) #last axis
```

##np.argsort()

```python
a7 = np.array([1,2,3,10,4,9])
print(np.argsort(a7))
print(np.argsort(a6))
```

```python
ar = np.array([20,10,40,30,50])
br = np.array([1,2,3,4,5])
print(ar, '\n', br)
```

```python
sorted = np.argsort(ar)
print(sorted)
```

```python
b_sorted = br[sorted]
print(b_sorted)
```

```python
ar1 = np.array([[1,21],[3,4],[5,9],[20,50],[30,20],[10,70]])
print(ar1)
```

```python
print(ar1[:,1])
print(ar1[:,0])
```

```python
idx = ar1[:,1].argsort()
print(idx)
```

```python
sorted_ar1_new = ar1[idx]
print(sorted_ar1_new)
```

##Boolean Filtering Mastery

```python
a = np.array([[[1,2,3,4],[5,6,7,8]],[[11,22,33,44],[55,66,77,88]]])
print(a)
```

```python
print(a[a>5])
```

##Multiple conditions

```python
print(a[(a>5)&(a<33)])
```

```python
print(a[(a>5) | (a<33)])

print(a[(a<5) | (a<8)])
```

##2D Filtering

```python
mask = a > 5
print(mask)
```

```python
masked = a[a> 5]
print(masked)
```

```python
masks  = a%2 == 0
print(masks)
a[masks] = 0
print(a)
```
