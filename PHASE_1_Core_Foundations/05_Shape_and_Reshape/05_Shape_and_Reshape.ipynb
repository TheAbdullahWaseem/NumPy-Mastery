```python
import numpy as np
```

```python
a = np.arange(12)
```

##Reshape

```python
b = a.reshape(4,3)
print(b)
```

##For 3D

```python
c = a.reshape(2,2,3)
print(c)
print(c.shape)
print(c.reshape(4,-1))
```

##Resize

```python
f = np.arange(16)
print(f)
```

```python
f.resize(4,4)
print(f)
```

```python
f.resize(5,4)
print(f)
```

##.flatten() (Always Copy)

```python
g = f.flatten()
print(g)
```

##.ravel() flatten but smart and faster

```python
a1 = np.arange(12)
a1.resize(5,5)
print(a1)
```

```python
a1.flatten()
```

```python
print(a1) #flatten returns copy
```

```python
flat = a1.ravel()
```

```python
print(a1)
print(flat) #returns view if possible
```

##.T (Transpose)

```python
h = np.arange(20)
h.resize(5,5)
print(h)
```

```python
t = h.transpose()
print(t)
```

##np.swapaxes()

```python
a3 = np.arange(36).reshape(3,4,3)
print(a3)
```

```python
a5 = np.swapaxes(a3,0,1)
print(a5)
```
