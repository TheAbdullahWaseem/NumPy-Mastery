Q4 (Tricky – Important Difference)

arr = np.arange(6)

b = arr.resize((3,2))

print(b)

print(arr)

What does b contain?

Why?

###Now try:

arr = np.arange(6)

b = np.resize(arr, (3,2))

print(b)

What is different?

Q6 (Tricky – Copy Behavior)

arr = np.arange(1,10).reshape(3,3)

b = arr.flatten()

b[0] = 999

print(arr)

```python
arr = np.arange(1,10).reshape(3,3)
b = arr.flatten()
b[0] = 999
print(arr)
print(b)
```

```python
np.shares_memory(arr, b) #It returns a copy
```

##RAVEL
🟢 Q7 (Simple)
arr = np.arange(1,10).reshape(3,3)

Ravel it.

```python
arr = np.arange(1,10).reshape(3,3)
ra = arr.ravel()
print(arr)
print(ra)
```

```python
np.shares_memory(arr,ra) #It returns a view
```

Q8 (Tricky – View vs Copy)

arr = np.arange(1,10).reshape(3,3)

b = arr.ravel()

b[0] = 999

print(arr)

Does original change?

##Now try:

arr = np.arange(1,10).reshape(3,3).T

b = arr.ravel()

b[0] = 999

print(arr)

Why is this behavior different?

```python
arr = np.arange(1,10).reshape(3,3)
b = arr.ravel()
b[0] = 999
print(arr)
np.shares_memory(arr,b) #It returns a view
```

```python
arr = np.arange(1,10).reshape(3,3).T
b = arr.ravel()
b[0] = 999
print(arr)
print(b)

'''
.T creates a non-contiguous array
Original memory order (C order):
1 2 3 4 5 6 7 8 9
After transpose:
It changes strides, not memory layout.

ravel() returns a view if possible otherwise returns a copy.
'''
```

Q12 (Tricky – Understanding Axes)

arr = np.arange(24).reshape(2,3,4)

b = np.swapaxes(arr, 1, 2)

What is the new shape?

If you modify b[0,0,0], does original change?

```python
arr = np.arange(24).reshape(2,3,4)
print(arr)
```

```python
b = np.swapaxes(arr,1,2)
print(b)
```

```python
b[0,0,0] = 100
print(arr) #Yes original changed so it is a view
```
