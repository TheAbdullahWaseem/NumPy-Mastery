##ndarray Object

```python
import numpy as np
```

```python
arr = np.array([[1,2,3],[4,5,6]])
new_arr = np.array([[11,12,13],[14,15,16]], dtype='float32')
print(arr)
print(new_arr)
```

##Important Attributes

```python
print("2 rows, 3 columns:",arr.shape)
print(arr.dtype)
print(arr.itemsize)
print(arr.size)
print(arr.nbytes)
print(arr.ndim)
```

```python
print("2 rows, 3 columns:",arr.shape)
print(new_arr.dtype)
print(new_arr.itemsize)
print(new_arr.size)
print(new_arr.nbytes)
print(new_arr.ndim)
```

```python
print(arr.flags)
```

##Upcasting

```python
#Change whole array to the that elements dtype that can be
#implemented to all the elements of array
ar = np.array([1,2,3,'string']) #convert all elements to string
print(ar.dtype)
a = np.array([1,2,3,3.5]) #convert all elements to float
print(a.dtype)
```
