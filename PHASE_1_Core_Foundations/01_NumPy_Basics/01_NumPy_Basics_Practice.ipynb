##View vs Copy

```python
arr = np.arange(10)
slice_arr = arr[2:7]
slice_arr[0] = 100
```

```python
print("arr: ",arr)
print("slice_arr: ",slice_arr)
```

```python
arr = np.array([1,2,3,4,5])
new_arr = arr.reshape(5,1)
new_arr[0] = 999
```

```python
print("arr: ",arr)
print("new_arr: ",new_arr)
```
