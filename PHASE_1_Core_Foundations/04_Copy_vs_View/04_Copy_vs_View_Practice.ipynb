```python
import numpy as np
```

Q1 — Basic Slicing (View or Copy?)

```python
arr = np.array([10, 20, 30, 40, 50])

b = arr[1:4]
b[0] = 999

print(arr)
```

```python
np.shares_memory(arr, b) #So it's a view as it shares memory
```

Q2 — Fancy Indexing Trap

```python
arr = np.array([10, 20, 30, 40, 50])

b = arr[[1, 2, 3]]
b[0] = 999
print(b)

print(arr)
```

```python
np.shares_memory(arr, b) #So it's a copy as it does not share memory
```

Q3 — Boolean Indexing Trap

```python
arr = np.array([5, 10, 15, 20])

b = arr[arr > 10]
b[0] = 999

print(arr)
```

```python
np.shares_memory(arr, b) #So it is also a copy
```

Q4 — 2D Slice + Modification

```python
arr = np.arange(1, 10).reshape(3,3)

b = arr[:, 1]
b[1] = 999


print(arr)
```

```python
np.shares_memory(arr, b) #It is a view
```

Q5 — Mixed Indexing (Very Tricky)

```python
arr = np.arange(1, 10).reshape(3,3)

b = arr[:, [1]]
b[1] = 999

print(arr)
```

```python
np.shares_memory(arr, b) #It is a copy
```

Q6 — .copy() Explicit

```python
arr = np.array([1,2,3,4,5])

b = arr[1:4].copy()
b[0] = 999

print(arr)
```

```python
np.shares_memory(arr, b) #It is a copy
```

Q7 — .view() Explicit

```python
arr = np.array([1,2,3,4,5])

b = arr.view()
b[0] = 999

print(arr)
```

```python
np.shares_memory(arr, b) #It is a view
```
