```python
import numpy as np
```

##View creates shallow copy

```python
arr = np.array([1,2,3,4])
v = arr.view()
v[0] = 100
print("Original", arr)
print("view", v)
```

##Copy creates deep copy

```python
c = arr.copy()
c[3] = 49
print("copy", c)
print("Original", arr)
```

##Slicing Behaviour

```python
a2 = np.array([1,2,3,4,5])
print("Original before slicing: ", a2)

s = a2[1:4]
s[2]= 100
print("Sliced", s)
print("Original after slicing", a2)
```

##Reshape Behaviour

```python
a3 = np.arange(6)
r = a3.reshape(2,3)
print("Original:", a3)
print("reshape: ",r)
r[1,2] = 100
print("\n")
print("Reshape after sliced", r)
print("Original also changed: ",a3)
```

##How to check if two arrays shre memory

```python
a4 = np.arange(6)
v2= a4.view()
c2 = a4.copy()

print(np.shares_memory(a4, v2))
print(np.shares_memory(a4, c2))
```
