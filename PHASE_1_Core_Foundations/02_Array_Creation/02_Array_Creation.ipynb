##np.array()

```python
import numpy as np
arr = np.array([1,2])
arr1 = np.array([[1,2],[3,4]])
print(arr,"\n", arr1)
```

##np.zeros()

```python
arr2 = np.zeros((2,3))
arr3 = np.zeros(5)
print(arr2)
print(arr3)
```

##np.ones()

```python
arr4 = np.ones((2,3), dtype = 'float32')
print(arr4)
```

##np.full()

```python
arr5 = np.full((2,3), 9)
print(arr5)
```

##np.full_like()

```python
arr6 = np.full_like((arr), 7, dtype = 'float')
print(arr6)
```

##np.arange()

```python
arr7 = np.arange(1,10)
arr8 = np.arange(1,10,2)
arr9 = np.arange(1,20,5)

print(arr7)
print(arr8)
print(arr9)
```

##np.linspace()

```python
arr10 = np.linspace(1,10,4)
arr11 = np.linspace(1,10,2, endpoint = False)
print(arr10)
print(arr11)
```

##np.eye() or np.identity()

```python
arr12 = np.eye(3)
arr13 = np.identity(4)
arr14 = np.eye(5, k=2)
print(arr12)
print(arr13)
print(arr14)
```

##np.random()

```python
a1 = np.random.rand(2,3)
print("rand:", a1)
a2 = np.random.randn(2,3)
print("randn: ",a2)
a3 = np.random.randint(1,10,(2,3))
print(a3)
a4 = np.random.choice([1,2,5,7], size= (2,3))
print(a4)
a5 = np.random.seed(42)
print(a5)
```

##np.empty()

```python
a6 = np.empty((2,3))
print(a6)
```

##dtype Control During Creation

```python
a7 = np.arange(0,10,3, dtype = 'float32')
print(a7)
```

```python
a8 = np.zeros((2,3), dtype = 'float32')
print(a8)
```
