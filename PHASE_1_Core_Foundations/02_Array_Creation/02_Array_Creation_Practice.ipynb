Q3 — arange vs linspace Precision

Create:

a = np.arange(0, 1, 0.2)
b = np.linspace(0, 1, 5)

Tasks:

Print both

Compare last element

```python
a = np.arange(0, 1, 0.2)
b = np.linspace(0, 1, 5)
print("arange: ",a) #last element will be excluded
print("linspace: ", b) #last element will be kept
```
